```python
import matplotlib.pyplot as plt
import seaborn as sns
import shap
```

```python
def set_visual_style():
    """Configura o estilo padrão dos gráficos para apresentações executivas."""
    sns.set_theme(style="whitegrid", palette="muted")
    plt.rcParams['figure.figsize'] = (10, 6)

def plot_confusion_matrix(matrix, title="Matriz de Confusão"):
    """Gera o mapa de calor da Matriz de Confusão."""
    plt.figure(figsize=(6, 4))
    sns.heatmap(matrix, annot=True, fmt='d', cmap='Blues')
    plt.title(title)
    plt.xlabel("Previsto (0=Abaixo da Meta, 1=Bateu Meta)")
    plt.ylabel("Real (0=Abaixo da Meta, 1=Bateu Meta)")
    plt.show()

def plot_shap_summary(model, X_test):
    """Gera o gráfico de explicabilidade do SHAP Values."""
    preprocessor = model.named_steps['preprocessor']
    rf_model = model.named_steps['classifier']
    
    X_transformed = preprocessor.transform(X_test)
    features = [nome.split('__')[-1] for nome in preprocessor.get_feature_names_out()]
    
    explainer = shap.TreeExplainer(rf_model)
    shap_values = explainer.shap_values(X_transformed)
    
    shap_vals_to_plot = shap_values[1] if isinstance(shap_values, list) else shap_values
    
    plt.figure()
    shap.summary_plot(shap_vals_to_plot, X_transformed, feature_names=features, show=False)
    plt.title("Impacto Global das Variáveis (SHAP Values)", fontsize=14)
    plt.tight_layout()
    plt.show()
```
