```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
```

```python
def evaluate_with_custom_threshold(model, X_test, y_test, threshold=0.55):
    """Avalia o modelo aplicando um limiar de confiança personalizado para mitigar Falsos Positivos."""
    y_proba = model.predict_proba(X_test)[:, 1]
    y_pred_custom = (y_proba >= threshold).astype(int)
    
    report = classification_report(y_test, y_pred_custom)
    matrix = confusion_matrix(y_test, y_pred_custom)
    
    return report, matrix
```
